```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內切換至 Google 小恐龍遊戲畫面！")
time.sleep(5)
print("小恐龍（對比度動態適應 - 1分鐘限時版）已啟動！")

start_script_time = time.time()

# ---------------- 參數設定區 (Mac 邏輯座標) ----------------
# 根據你的參數優化，並放大 2 倍對應 Mac Retina 物理像素
BASE_X = 120          # 恐龍前方偵測距離基準
SEARCH_WIDTH = 40     # 偵測區域寬度

# 1. 低空/地面跳躍區
JUMP_Y_TOP = 295      
JUMP_HEIGHT = 40      

# 2. 中空翼龍區
MID_PTERO_Y_TOP = 265   
MID_PTERO_HEIGHT = 30   
# --------------------------------------------------------

def has_obstacle_contrast(im, width, height):
    """
    透過對比度判斷障礙物：
    先抓取區塊左上角 (0,0) 作為【背景基準色】，
    只要區塊內有像素與背景色的 RGB 差異超過 50，即視為障礙物！
    """
    bg_r, bg_g, bg_b = im.getpixel((0, 0))
    
    # 為了效能流暢，步長維持 4 像素（在 Mac 2倍放大後等同 8 像素精細度）
    for x in range(0, width, 4):
        for y in range(0, height, 4):
            r, g, b = im.getpixel((x, y))
            if abs(r - bg_r) > 50 or abs(g - bg_g) > 50 or abs(b - bg_b) > 50:
                return True
    return False

try:
    while True:
        # 【考試必備】：穩定執行滿 60 秒（1 分鐘）自動停止
        if time.time() - start_script_time > 60:
            print("已執行滿 1 分鐘，自動停止腳本。")
            break

        elapsed_time = time.time() - start_script_time
        speed_offset = int(min(elapsed_time * 0.8, 60))
        
        # 動態計算邏輯座標，並乘以 2 完美適配 Mac Retina 螢幕截圖
        current_logic_x = BASE_X + speed_offset
        
        mac_x = current_logic_x * 2
        mac_jump_y = JUMP_Y_TOP * 2
        mac_jump_w = SEARCH_WIDTH * 2
        mac_jump_h = JUMP_HEIGHT * 2
        
        mac_mid_y = MID_PTERO_Y_TOP * 2
        mac_mid_h = MID_PTERO_HEIGHT * 2

        # 1. 擷取低空/地面跳躍區 (Mac 雙倍像素)
        im_jump = pyg.screenshot(region=(mac_x, mac_jump_y, mac_jump_w, mac_jump_h))
        
        # 2. 擷取中空翼龍區 (Mac 雙倍像素)
        im_mid = pyg.screenshot(region=(mac_x, mac_mid_y, mac_jump_w, mac_mid_h))

        should_jump = False
        should_duck = False

        # A. 檢查地面區是否有對比度異常
        if has_obstacle_contrast(im_jump, mac_jump_w, mac_jump_h):
            should_jump = True

        # B. 地面安全時，檢查中空區是否有對比度異常
        elif has_obstacle_contrast(im_mid, mac_jump_w, mac_mid_h):
            should_duck = True

        # 執行動作與俐落的冷卻時間
        if should_jump:
            pyg.press('space')
            time.sleep(0.12)
        elif should_duck:
            pyg.keyDown('down')
            time.sleep(0.2)
            pyg.keyUp('down')
            
        time.sleep(0.005)

except KeyboardInterrupt:
    print("程式已停止。")
```

```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內切換至 Google 小恐龍遊戲畫面！")
time.sleep(5)
print("小恐龍（座標完美對齊實測版）已啟動！")

start_script_time = time.time()

# ---------------- 參數設定區 (對應你剛剛 MouseInfo 實測數值) ----------------
BASE_X = 310          # 恐龍前方偵測距離基準（完美對齊你的 X:97）
SEARCH_WIDTH = 60     # 偵測區域寬度（稍微加寬，確保高矮仙人掌都涵蓋）

# 1. 低空/地面跳躍區（完美對齊你的 Y:425 地面高度）
JUMP_Y_TOP = 600                 
JUMP_HEIGHT = 40        

# 2. 中空翼龍區（保持安全高度，不干擾）
MID_PTERO_Y_TOP = 360   
MID_PTERO_HEIGHT = 30   
# --------------------------------------------------------

def has_obstacle_contrast(im, width, height):
    """
    透過像素密度判斷障礙物：
    累積足夠的深色像素才觸發，過濾掉碎石雜訊
    """
    dark_pixel_count = 0
    
    for x in range(0, width, 4):
        for y in range(0, height, 4):
            pixel = im.getpixel((x, y))
            r, g, b = pixel[0], pixel[1], pixel[2]
            
            if r < 200 and g < 200 and b < 200:
                dark_pixel_count += 1
                
    # 超過 10 個深色像素就判定為障礙物
    if dark_pixel_count > 10:
        return True
        
    return False

try:
    while True:
        elapsed_time = time.time() - start_script_time
        speed_offset = int(min(elapsed_time * 0.4, 30))
        
        current_logic_x = BASE_X + speed_offset
        
        mac_x = current_logic_x * 2
        mac_jump_y = JUMP_Y_TOP * 2
        mac_jump_w = SEARCH_WIDTH * 2
        mac_jump_h = JUMP_HEIGHT * 2
        
        mac_mid_y = MID_PTERO_Y_TOP * 2
        mac_mid_h = MID_PTERO_HEIGHT * 2

        # 1. 擷取低空/地面跳躍區 (Mac 雙倍像素)
        im_jump = pyg.screenshot(region=(mac_x, mac_jump_y, mac_jump_w, mac_jump_h))
        
        # 2. 擷取中空翼龍區 (Mac 雙倍像素)
        im_mid = pyg.screenshot(region=(mac_x, mac_mid_y, mac_jump_w, mac_mid_h))

        should_jump = False
        should_duck = False

        # A. 檢查地面區是否有障礙物
        if has_obstacle_contrast(im_jump, mac_jump_w, mac_jump_h):
            should_jump = True

        # B. 檢查中空區是否有障礙物
        elif has_obstacle_contrast(im_mid, mac_jump_w, mac_mid_h):
            should_duck = True

        # 執行動作與俐落的冷卻時間
        if should_jump:
            pyg.press('space')
            time.sleep(0.25)     # 扎實的冷卻時間，避免空中連跳
        elif should_duck:
            pyg.keyDown('down')
            time.sleep(0.2)
            pyg.keyUp('down')
            
        time.sleep(0.01)

except KeyboardInterrupt:
    print("程式已停止。")
```

```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內點擊遊戲畫面！")
pyg.sleep(5)

print("小恐龍自動遊玩（終極穩定連續版）已啟動！")

last_jump_time = 0

try:
    while True:
        # 冷卻保護：跳完後 0.38 秒內不重複觸發，徹底解決卡住不動的問題
        if time.time() - last_jump_time < 0.38:
            time.sleep(0.005)
            continue

        im = pyg.screenshot()
        obs = False

        # 結合你先前成功會跳的黃金基準
        start_x = 310  
        start_y = 580  # 從較高處開始，確保高仙人掌頂部不漏接

        for x in range(start_x, start_x + 70, 4):
            # 垂直掃描 240 像素（從 580 掃到 820），高矮仙人掌通吃
            for y in range(start_y, start_y + 240, 5):
                
                # 關鍵：直接用 (x, y)，不需要乘以 2！
                c = im.getpixel((x, y))
                
                # 偵測深色障礙物 (RGB 低於 100)
                if c[0] < 100 and c[1] < 100 and c[2] < 100:
                    obs = True
                    break
            if obs:
                break
                
        if obs:
            print(">>> 發現障礙物！起跳！")
            
            pyg.keyDown('space')
            time.sleep(0.18)
            pyg.keyUp('space')
            
            # 記錄起跳時間，進入冷卻計時，保證下一個障礙物還能繼續跳
            last_jump_time = time.time()
            
        time.sleep(0.003)

except KeyboardInterrupt:
    pyg.keyUp('space')
    print("程式已停止。")
```

```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內點擊遊戲畫面！")
pyg.sleep(5)

print("小恐龍自動遊玩（高矮通吃精準版）已啟動！")

was_obs = False

try:
    while True:
        # 使用全螢幕截圖（保持你這個最順利運作的架構）
        im = pyg.screenshot()
        obs = False

        # 【高矮通吃座標設定】
        start_x = 310  
        start_y = 570  # 稍微往上拉，確保高仙人掌的頭頂不漏接

        # 將 X 寬度加寬到 90，Y 範圍拉長到 270（同時涵蓋高與矮）
        for x in range(start_x, start_x + 90, 4):
            for y in range(start_y, start_y + 270, 5):

                c = im.getpixel((x, y))

                # 偵測深色障礙物 (RGB 低於 100)
                if c[0] < 100 and c[1] < 100 and c[2] < 100:
                    obs = True
                    break

            if obs:
                break

        # 1. 當發現障礙物且之前尚未起跳時，執行跳躍
        if obs and not was_obs:
            print("偵測到障礙物（高/矮）！跳躍！")

            pyg.keyDown('space')
            time.sleep(0.18)
            pyg.keyUp('space')

            was_obs = True  # 標記「已經起跳」

        # 2. 當畫面上的障礙物通過後（obs 變回 False），立刻重設旗標！
        if not obs:
            was_obs = False

        time.sleep(0.003)

except KeyboardInterrupt:
    pyg.keyUp('space')
    print("程式已停止。")
```
