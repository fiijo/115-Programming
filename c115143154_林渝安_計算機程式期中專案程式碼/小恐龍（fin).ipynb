```python
import pyautogui
import time

print("3 秒後開始，請切換到恐龍遊戲...")
time.sleep(3)

# 偵測區域
x1 = 100
y1 = 400
x2 = 500
y2 = 550

while True:

    screenshot = pyautogui.screenshot(
        region=(x1, y1, x2 - x1, y2 - y1)
    )

    pixels = screenshot.load()
    obstacle = False

    for x in range(screenshot.width):
        for y in range(screenshot.height):

            r, g, b = pixels[x, y]

            if r < 100 and g < 100 and b < 100:
                obstacle = True
                break

        if obstacle:
            break

    if obstacle:
        print("偵測到障礙物！跳躍！")
        pyautogui.press("space")
        time.sleep(0.3)

    time.sleep(0.02)

```

```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內點擊遊戲畫面！")
pyg.sleep(5)

print("小恐龍自動遊玩（全螢幕連續跳躍修正版）已啟動！")

was_obs = False

try:
    while True:
        # 使用全螢幕截圖（不加 region，徹底避免 Mac 邊界報錯）
        im = pyg.screenshot()
        obs = False

        # 保留你實測最精準的偵測範圍與點位
        start_x = 310
        start_y = 600

        for x in range(start_x, start_x + 80, 4):
            for y in range(start_y, start_y + 240, 5):

                c = im.getpixel((x, y))

                if c[0] < 100 and c[1] < 100 and c[2] < 100:
                    obs = True
                    break

            if obs:
                break

        # 1. 當發現障礙物且之前尚未起跳時，執行跳躍
        if obs and not was_obs:
            print("偵測到障礙物！跳躍！")

            pyg.keyDown('space')
            time.sleep(0.18)
            pyg.keyUp('space')

            was_obs = True  # 標記「已經起跳」

        # 2. 當畫面上的障礙物通過後（obs 變回 False），立刻重設旗標！
        # 這樣下一個仙人掌出現時，就能順利進行第二次、第三次跳躍！
        if not obs:
            was_obs = False

        time.sleep(0.003)

except KeyboardInterrupt:
    pyg.keyUp('space')
    print("程式已停止。")
```

```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內點擊遊戲畫面！")
pyg.sleep(5)
print("小恐龍（無限暢玩・Mac 完美修復版）已啟動！")

try:
    while True:
        # 已移除一分鐘限制，可以無限暢玩！

        im = pyg.screenshot()
        obs = False
        
        start_x = 310  
        start_y = 660  
        
        # 網格掃描
        for x in range(start_x, start_x + 60, 4):
            for y in range(start_y, start_y + 180, 5):
                
                # 【關鍵修正】：Mac Retina 截圖尺寸是兩倍，必須乘以 2 才能精準對應畫面位置！
                c = im.getpixel((x * 2, y * 2))
                
                # 偵測深色障礙物 (RGB 低於 100)
                if c[0] < 100 and c[1] < 100 and c[2] < 100:
                    obs = True
                    break
            if obs:
                break
                
        if obs:
            print(">>> 偵測到障礙物！起跳！")
            pyg.press('space')
            # 加上適當的冷卻時間，避免在空中連續按空白鍵
            time.sleep(0.2)  
            
        time.sleep(0.003)

except KeyboardInterrupt:
    print("程式已停止。")
```
