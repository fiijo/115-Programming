```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內點擊遊戲畫面並開始遊戲！")
pyg.sleep(5)

print("小恐龍自動遊玩（滑鼠實測精準版）已啟動！")

last_jump_time = 0

# 【使用滑鼠親自定位的實體絕對座標】
start_x = 84 * 2    # 168
start_y = 426 * 2   # 852
scan_width = 40 * 2   # 80
scan_height = 20 * 2  # 40

# 鎖定偵測區域的矩形邊界 (left, upper, right, lower)
box = (start_x, start_y, start_x + scan_width, start_y + scan_height)

try:
    while True:
        # 冷卻保護
        if time.time() - last_jump_time < 0.25:
            time.sleep(0.002)
            continue

        im = pyg.screenshot()
        
        # 瞬間把偵測區塊整塊切出來
        region = im.crop(box)
        
        # 取得該區域內 RGB 各通道的最小值
        extrema = region.getextrema()
        min_r = extrema[0][0]
        min_g = extrema[1][0]
        min_b = extrema[2][0]

        # 只要區域內有任何深色像素（低於 100），立刻起跳！
        if min_r < 100 and min_g < 100 and min_b < 100:
            print(f">>> 🎯 瞬間抓到障礙物！起跳！(RGB極值: {min_r}, {min_g}, {min_b})")
            
            pyg.keyDown('space')
            time.sleep(0.15)
            pyg.keyUp('space')
            
            last_jump_time = time.time()

        time.sleep(0.002)

except KeyboardInterrupt:
    pyg.keyUp('space')
    print("程式已停止。")
```

```python
import time
import pyautogui as pyg

print("🚀 【防誤判・精準仙人掌版本】啟動！")
print("請在 5 秒內點擊一下遊戲畫面...")
time.sleep(5)

BASE_X = 79           
SEARCH_WIDTH = 45     

# 關鍵調整：把 Y 往上提一點（離開地平線雜訊），高度縮小
JUMP_Y_TOP = 412      
JUMP_HEIGHT = 20      

mac_x = BASE_X * 2
mac_jump_y = JUMP_Y_TOP * 2
mac_jump_w = SEARCH_WIDTH * 2
mac_jump_h = JUMP_HEIGHT * 2

try:
    while True:
        im = pyg.screenshot().convert('RGB')
        region = im.crop((mac_x, mac_jump_y, mac_x + mac_jump_w, mac_jump_y + mac_jump_h))
        
        bg_r, bg_g, bg_b = region.getpixel((0, 0))
        
        max_diff = 0
        for x in range(0, mac_jump_w, 4):
            for y in range(0, mac_jump_h, 4):
                r, g, b = region.getpixel((x, y))
                diff = max(abs(r - bg_r), abs(g - bg_g), abs(b - bg_b))
                if diff > max_diff:
                    max_diff = diff

        # 提高門檻到 75，確保只有真正的仙人掌靠近時才跳
        if max_diff > 75:
            print(f"🎯 抓到仙人掌！色差值: {max_diff}")
            pyg.press('space')
            time.sleep(0.12)  # 起跳冷卻
            
        time.sleep(0.01)

except KeyboardInterrupt:
    print("程式已停止。")
```

```python
import time
import pyautogui as pyg

print("🚀 【動態速度追蹤・完美時機自動通關】啟動！")
print("請在 5 秒內點擊一下 Chrome 的遊戲畫面...")
for i in range(5, 0, -1):
    print(f"倒數 {i} 秒...")
    time.sleep(1)

print("⚡ 開始自動通關！")

start_script_time = time.time()

# 基礎參數
BASE_X = 180          # 初始預判距離
SEARCH_WIDTH = 50     
JUMP_Y_TOP = 412      
JUMP_HEIGHT = 20      

try:
    while True:
        # 計算遊戲進行了多久，隨著時間讓偵測點微微向右延伸（適應遊戲加速）
        elapsed_time = time.time() - start_script_time
        dynamic_base_x = BASE_X + int(min(elapsed_time * 0.5, 40))
        
        mac_x = dynamic_base_x * 2
        mac_jump_y = JUMP_Y_TOP * 2
        mac_jump_w = SEARCH_WIDTH * 2
        mac_jump_h = JUMP_HEIGHT * 2

        im = pyg.screenshot().convert('RGB')
        region = im.crop((mac_x, mac_jump_y, mac_x + mac_jump_w, mac_jump_y + mac_jump_h))
        
        has_obstacle = False
        for x in range(0, mac_jump_w, 6):
            for y in range(0, mac_jump_h, 6):
                r, g, b = region.getpixel((x, y))
                if r < 200 or g < 200 or b < 200:
                    has_obstacle = True
                    break
            if has_obstacle:
                break

        if has_obstacle:
            pyg.press('space')
            time.sleep(0.25)  # 完美的起跳冷卻，不卡連跳
            
        time.sleep(0.001)

except KeyboardInterrupt:
    print("程式已停止。")
```
