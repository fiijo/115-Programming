```python
import time
import pyautogui as pyg

# 預先等待 2 秒讓你切換回遊戲畫面
pyg.sleep(2)

print("小恐龍自動遊玩程式已啟動！按 Ctrl+C 可以停止。")

# 主迴圈
try:
    while True:
        # 建議將 region 調整為恐龍前方一小條區域 (x, y, 寬度, 高度)
        # 這裡以畫面實際座標為主，你可以用 MouseInfo 微調這組區域
        # region=(起始X, 起始Y, 寬度, 高度)
        
        # 取得畫面截圖中的一小塊前方區域
        # 註：請確保座標對應到你瀏覽器中仙人掌會經過的路徑
        im = pyg.screenshot(region=(90, 435, 50, 30)) 
        
        obs = False
        
        # 檢查這個區域內是否有接近障礙物的深色像素（仙人掌或小鳥）
        # 迴圈掃描區域內的像素顏色
        for x in range(0, 60, 5):  # 每隔 5 個像素掃描一次以加快速度
            for y in range(0, 30, 5):
                c = im.getpixel((x, y))
                
                # 判斷是否為障礙物顏色（Chrome 預設日間模式障礙物約為深灰色/黑色 RGB 83,83,83 左右）
                # 如果背景是深色模式，這裡的顏色判斷條件需要改成白色 (255,255,255)
                if c[0] < 100 and c[1] < 100 and c[2] < 100:  # 偵測暗色系像素
                    obs = True
                    break
            if obs:
                break
                
        if obs:
            pyg.press('space')
            time.sleep(0.08)  # 跳躍後的短暫緩衝，避免重複連跳
            
        # 稍微暫停以免 CPU 佔用過高
        time.sleep(0.01)                                            

except KeyboardInterrupt:                 
    print("程式已停止。")
```

```python
import time
import pyautogui as pyg

pyg.sleep(2)
print("開始偵測... 請確保滑鼠已點擊 Chrome 遊戲畫面！")

try:
    while True:
        # 💡 將 X 設為 140（在恐龍右側前方），讓它去偵測右邊過來的仙人掌
        # region=(起始X, 起始Y, 寬度, 高度)
        im = pyg.screenshot(region=(90, 435, 50, 30))
        obs = False
        
        # 掃描區域內的像素
        for x in range(0, 50, 10):
            for y in range(0, 30, 10):
                c = im.getpixel((x, y))
                
                # 偵測深色障礙物 (RGB 低於 100)
                if c[0] < 100 and c[1] < 100 and c[2] < 100:
                    obs = True
                    break
            if obs:
                break
                
        if obs:
            print("偵測到障礙物！起跳！")
            pyg.press('space')
            time.sleep(0.25)  # 跳躍後的緩衝時間
            
        time.sleep(0.01)

except KeyboardInterrupt:
    print("程式已停止。")
```
