```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內點擊遊戲畫面！")
pyg.sleep(5)

print("小恐龍自動遊玩（高矮通吃精準版）已啟動！")

was_obs = False

try:
    while True:
        # 使用全螢幕截圖（保持你這個最順利運作的架構）
        im = pyg.screenshot()
        obs = False

        # 【高矮通吃座標設定】
        start_x = 95  
        start_y = 410  # 稍微往上拉，確保高仙人掌的頭頂不漏接

        # 將 X 寬度加寬到 90，Y 範圍拉長到 270（同時涵蓋高與矮）
        for x in range(start_x, start_x + 90, 4):
            for y in range(start_y, start_y + 270, 5):

                c = im.getpixel((x, y))

                # 偵測深色障礙物 (RGB 低於 100)
                if c[0] < 100 and c[1] < 100 and c[2] < 100:
                    obs = True
                    break

            if obs:
                break

        # 1. 當發現障礙物且之前尚未起跳時，執行跳躍
        if obs and not was_obs:
            print("偵測到障礙物（高/矮）！跳躍！")

            pyg.keyDown('space')
            time.sleep(0.18)
            pyg.keyUp('space')

            was_obs = True  # 標記「已經起跳」

        # 2. 當畫面上的障礙物通過後（obs 變回 False），立刻重設旗標！
        if not obs:
            was_obs = False

        time.sleep(0.003)

except KeyboardInterrupt:
    pyg.keyUp('space')
    print("程式已停止。")
```

```python
import pyautogui as pyg

im = pyg.screenshot()

# 【黃金微調：向下貼近地面，但避開地板雜訊】
# 1. X 保持安全預判距離：95 * 2 = 190
# 2. Y 往下調到 412 (實體 824)，讓紅框剛好落在仙人掌會經過的中下段！
# 3. 高度設定為 20 (實體 40)，精準鎖定障礙物本體
test_x = 95 * 2    # 190
test_y = 412 * 2   # 824

scan_w = 40 * 2   
scan_h = 20 * 2   

for x in range(test_x, test_x + scan_w):
    for y in range(test_y, test_y + scan_h):
        if x < im.width and y < im.height:
            im.putpixel((x, y), (255, 0, 0))  # 漆成紅色方框

im.save("debug_box.png")
print("✅ 微調紅框已儲存！快打開 'debug_box.png' 看看這次的高度是不是剛好卡在仙人掌的高度位置？")
```

```python
import time
import pyautogui as pyg

print("自動遊玩準備中... 請在 5 秒內點擊遊戲畫面！")
pyg.sleep(5)

print("小恐龍自動遊玩（黃金甜蜜點通關版）已啟動！")

last_jump_time = 0

try:
    while True:
        # 冷卻保護：跳完後 0.35 秒內不重複觸發
        if time.time() - last_jump_time < 0.35:
            time.sleep(0.005)
            continue

        im = pyg.screenshot()
        obs = False

        # 【黃金絕對座標】：精準鎖定仙人掌必經中段！
        start_x = 95 * 2    # 190 (實體像素)
        start_y = 412 * 2   # 824 (實體像素)

        # 掃描偵測範圍
        scan_width = 40 * 2   
        scan_height = 20 * 2   

        for x in range(start_x, start_x + scan_width, 4):
            for y in range(start_y, start_y + scan_height, 4):

                c = im.getpixel((x, y))

                # 偵測深色障礙物 (RGB 低於 100)
                if c[0] < 100 and c[1] < 100 and c[2] < 100:
                    obs = True
                    break

            if obs:
                break

        # 偵測到障礙物時立刻起跳
        if obs:
            print(">>> 🎯 完美命中仙人掌！起跳！")

            pyg.keyDown('space')
            time.sleep(0.15)  # 按壓時間
            pyg.keyUp('space')

            last_jump_time = time.time()

        time.sleep(0.002)

except KeyboardInterrupt:
    pyg.keyUp('space')
    print("程式已停止。")
```

```python
import time
import pyautogui as pyg

print("請在 5 秒內點擊遊戲並讓它開始跑，我們來看看程式實際「看」到什麼顏色！")
pyg.sleep(5)

print("開始檢測偵測區塊的顏色...")

try:
    for _ in range(20):  # 連續檢測 20 次
        im = pyg.screenshot()
        
        start_x = 95 * 2    # 190
        start_y = 412 * 2   # 824
        scan_width = 40 * 2   
        scan_height = 20 * 2   

        min_val = 255
        for x in range(start_x, start_x + scan_width, 6):
            for y in range(start_y, start_y + scan_height, 6):
                c = im.getpixel((x, y))
                # 取 RGB 中最小的值（看有沒有接近黑色）
                val = min(c[0], c[1], c[2])
                if val < min_val:
                    min_val = val

        print(f"👀 目前偵測區塊內抓到的最低像素亮度 (0=純黑, 255=純白): {min_val}")
        time.sleep(0.5)
        
    print("檢測結束！")
    
except KeyboardInterrupt:
    print("已手動停止。")
```
