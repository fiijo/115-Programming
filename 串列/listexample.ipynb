# 簡介
串列(list)是一個可以儲存多個元素(整數，浮點數，或字串)的容器。
串列中每一個元素都是使用逗號隔開，左右有中括號[]。

```python
score=[78,29,56,80,99]
print(score)
```

# 索引值
我們可以使用索引值(index)來儲存串列中的元素。
注意:索引值由0開始。

```python
s=[78,29,56,80,99]
print(s[0])
print(s[1])
```

## 串列內容的個數
len(a) 取得a這個串列中元素的個數。
len 是 length(長度) 的意思

```python
print(len(s))
print(s[-1])
print(s[len(s)-1])
```
