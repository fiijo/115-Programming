# 簡介
串列(list)是一個可以儲存多個元素(整數，浮點數，或字串)的容器。
串列中每一個元素都是使用逗號隔開，左右有中括號[]。

```python
score=[78,29,56,80,99]
print(score)
```

# 索引值
我們可以使用索引值(index)來儲存串列中的元素。
注意:索引值由0開始。

```python
s=[78,29,56,80,99]
print(s[0])
print(s[1])
```

## 串列內容的個數
len(a) 取得a這個串列中元素的個數。
len 是 length(長度) 的意思

```python
print(len(s))
print(s[-1])
print(s[len(s)-1])
```

## 新增元素
使用`append()` 新增元素到串列的尾端

```python
S=[78,29,56,80,99]
S.append(100)
print(S)
S.append(102)
print(S)
```

```python
print(s)
s[2]=70
print(s)
```

# 走訪串列的三種寫法


```python
print(s)
for i in range(len(s)):
    print(s[i])
```

```python
#對於s中的每個元素，一次取一個，放在e中。
for e in s:
    print(e)
```

```python
for i, e in enumerate(s):
    print(i,e)
```

## 計算費伯納西數列的第 91 個元素
一個費伯納西數列如下:
1,1,2,3,5,8,.....
請問第91個元素為何?  
提示:  
令$x$是一個費伯納西數列 $x[i]$ 是第 $i$個數字。     
我們會得到 $x[n]$=$x[n-1]$+$x[n-2]$

```python
fib=[0]*91
print(fib)
fib[0]=1
fib[1]=1
print(fib)
i=2
fib[i]=fib[i-2]+fib[i-1]
print(fib)
i=3
fib[i]=fib[i-2]+fib[i-1]
print(fib)
i=4
fib[i]=fib[i-2]+fib[i-1]
print(fib)
```

```python
for i in range(2,91):
    fib[i]=fib[i-2]+fib[i-1]
print(fib[90])
```
